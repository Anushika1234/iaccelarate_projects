```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.mixture import GaussianMixture
from sklearn.cluster import AgglomerativeClustering

# Load your dataset
df = pd.read_csv("period - Copy.csv")  # Update with your file path

# Convert all columns to numeric (ensuring no string values remain)
df = df.apply(pd.to_numeric, errors='coerce')

# Handle missing values (use mean/mode/median)
imputer = SimpleImputer(strategy="median")  # You can also use "mean" or "most_frequent"
df_imputed = imputer.fit_transform(df)

# Scale the dataset
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_imputed)

# Confirm the dataset is valid
assert df_scaled.shape[0] > 0, "Dataset is empty after preprocessing!"
assert not np.isnan(df_scaled).any(), "NaN values detected in df_scaled!"

# Train models
hierarchical = AgglomerativeClustering(n_clusters=3, linkage='ward')
hierarchical.fit(df_scaled)

gmm = GaussianMixture(n_components=3, random_state=42)
gmm.fit(df_scaled)

print("Clustering models trained successfully!")
```
